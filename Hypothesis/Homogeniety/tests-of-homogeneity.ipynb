# Calculating the expectancy table

The null hypothesis for the test of homogeneity that we are learning about in this series of exercises is that all the distributions are identical.  If this hypothesis is true then:

$$
P(X=i) \approx \frac{ \sum_{k=1}^r C_{ik} }{ \sum_{k=1}^c \sum_{l=1}^r C_{kl} } 
$$

is a maximum likelihood estimator for the probability that the random variable is equal to $i$. To understand why consider what the terms in this expression mean. $C_{ij}$, is the matrix that you generated in the last exercise. $C_{ij}$ is, therefore, the number of appearances of the integer $i$ in the $j$th sample of data. Consequently, the summation in the numerator of the above expression is the total number of appearance of $i$ in all $r$ samples. The denominator is then the total number of data points in all $r$ samples.  The quotient above is thus the fraction of the random variables in all $r$ samples that are equal to $i$, which is the maximum likelihood estimator for the probability of the event $X=i$.  

Note that the above is only the maximum likelihood estimator if all $r$ distributions are identical. In other words, we can only use the above expression to estimate the probability that $P(X=i)$ if the null hypothesis is true.  However, if the null hypothesis is true we can also estimate the expected value for the number of appearances of the value $i$ in the $j$th of our samples. This quantity is given by:

$$
E_{ij} = \frac{ \sum_{k=1}^r C_{ik} }{ \sum_{k=1}^c \sum_{l=1}^r C_{kl} } \sum_{l=1}^c C_{lj}
$$

Notice that the quotient in this expression is the maximum likelihood estimator for the probability that was discussed above. This probability can be thought of as the $p$ parameter for a Bernoulli random variable.  The second summation is then the number of samples that were used to contruct our $j$th estimate of the histogram. We are thus multiplying our maximum likelihood estimator for the probability of observing an $i$ by the total number of opportunities that there were to generate an $i$ in the $j$th sample.  In other words, we are calculating the expectation of a sum of $n$ Bernoulli random variables as $np$, where $n$ is the number of samples of the Bernoulli random variable.
