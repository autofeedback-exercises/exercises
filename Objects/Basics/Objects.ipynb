As you can see, __for python__, a __set__ is an object that contains a set of __distinguishable__ objects. It thus has a finite __cardinality__ (the number of objects in the set) and you can __iterate__ over the elements in the __set__. If you make the individual elements in the set __addressable__ then it becomes a __tuple__.  A __tuple__  thus has a finite __cardinality__, you can __iterate__ over its elements and you can __address__ each of the elements. However, the elements of a __tuple__ are __not mutable__ in the way that the elements of a __list__ are. If a __list__ is defined in a python code then it has a __cardinality__, you can __iterate__ over its elements, you can __address__ each element and you can also __mutate__ (i.e. change) the elements of the list later in your code. When you do arithmetic operations upon it (e.g. addition, subtraction, multiplication and division) it does not behave as a __vector__ would.  Thankfully, however, NumPy arrays will behave like vectors when you add or subtract pairs of them and when you multiply or divide them by scalars.

# An aside on the need for v1.copy() in the code above

Some of you may be asking what the `v1.copy()` in the previous function is for.  To understand why we need this consider the following code:

```python
v1 = [1,2,3]
v2 = v1
v2[1] = 1
print( "V1 is", v1, "V2 is", v2 )
```

If you think that this will print out:

````
V1 is [1, 2, 3] V2 is [1, 1, 3]
````

you are due a horrible surprise as it will actually print:

````
V1 is [1, 1, 3] V2 is [1, 1, 3]
````

In other words, `v1` will be changed even though the code only appears to be changing `v2`. This is because `v2` is a shallow copy of `v1`.  We use `v1.copy()` to make `v2` a deep copy of `v1`.  Using `v1.copy()` when making `v2` ensures that we can change the elements of `v2` __without__ changing the elements of `v1`.

In the last exercise you saw how we can use a __dictionary__ to store the properties for each of the objects in a __set__.  In this final exercise, we will think about an object that can be used to hold the properties of the cartesian product of two __sets__.

The squares in this figure are coloured in accordance with the cardinality of the __intersection__ between the __set__ of solutions to $x^n-1=0$ and the __set__ of solutions to $x^m-1=0$.  If square $n, m$ in figure 3 has a light colour it thus means that there are many solutions for $x^n-1=0$ that are also solutions to $x^m-1=0$.  

Lets introduce some symbols to understand how this figure was constructed.

* $E_N$ symbolises the __set__ of equations of the form $z^n-1=0$ with $n \in \mathbb{N}$ and $n\le N$.
* $S_n$ is the __set__ of solutions to $x^n-1=0$

We are taking the cartesian product of $E_N$ with itself (i.e. $E_N \times E_N$). The $n,m$ element of this cartesian product is thus the __tuple__ of equations $(x^n-1=0,x^m-1=0)$.  From each of these __tuples__ of equations we can construct a __tuple__ of __sets__ $(S_n,S_m)$ by solving for $x$ using sympy. We can then determine the __intersection__ the two sets in the __tuple__ $S_n \cap S_m$.  The final quantity that we interseted in is the __cardinality__ of this __intersection__ $|S_n \cap S_m|$. 

To simplify matters we can label each of these __cardinalities__ of an __intersection__ with a __tuple__ of natural numbers, i.e. the values of $n$ and $m$ in the underlying __tuple__ of equations.  Doing so means that we can use a __matrix__ (or 2D NumPy __array__) to hold all the cardinalities.  To create the required matrix using python we would write:

```python
mymatrix = np.zeros([N,N])
```

```python

```
