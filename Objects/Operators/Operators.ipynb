# Introduction

In the first of these exercises on objects in mathematics you saw how we can use mathematical objects represent elements in the world. In particular, we saw that:

* We can use a __set__ to hold collections of distinguishable things.
* We can use a __number__ to describe a property of a thing.
* We can use a __vector__ to describe the properties of all the things in a __set__
* We can use a __matrix__ to describe the properties of all the pairs of things in one __set__ or a pair of __sets__.

If that were all we could do with these mathematical objects, mathematics would be far less significant. However, mathematics is useful because we can apply operations to these various objects independently of the elements of the world that they represent. This fact is important because it means that we can study __sets__, __numbers__, __vectors__ and __matrices__ in the abstract. In other words, we can do mathematics on __sets__, __numbers__, __vectors__ and __matrices__ that are not representative of anything.  For example we can make the statement:

$$
17+6 = 23
$$

without knowing what there are 17 and 6 of. However, to understand what this means we need to know a great number of things. For example, we need to know that: 

* numbers can be used to represent the cardinality of sets.
* How to count to 23.
* What addition means.
* What the equals sign means.

Furthermore, we also need to remember that there is no guarantee that the cardinality of the union, $A \cup B$ of two sets, $A$ and $B$, with cardinalities of 17 and 6 is 23.  If set $A$ contains elements that are also in $B$ then the cardinality of the union will be less than 23.  All this complexity is why Bertrand Russell famously wrote a proof that $1+1=2$ that is 362 pages long! 

However, mathematics allows us to retreat even further into abstraction by writing statements such as:

> If $a$ is an odd number and $b$ is an even number then $a+b$ must be odd.

This statement is more abstract because we know don't even what numbers $a$ and $b$ represent. It is also harder to understand because to parse it we need to know all the things from the previous list as well as

* How odd and even numbers are defined.
* That we are using the symbols $a$ and $b$ to represents numbers.
* The meaning of every english word in the above sentence

In short, this stuff is complicated. It doesn't feel particularly complicated because it is now familiar. However, when you and your colleagues first learned this stuff in primary school you almost certainly found it quite hard. The fact that you understood it all is a testament to the amount of time your teachers gave you to mess around with __the process__ of adding numbers together. This messing about is what provided you with the __fluency__ that is required to understand statements, like the one above, that are about the outcome of performing this operation in the abstract. 

The problem for students learning about mathematics at university is that we want you to quickly go from learning about a complicated __process__ (e.g. multiplying two matrices together or drawing a graph of a function) to having the __fluency__ required to understand statements about the outcomes of performing these complicated process in the abstract. It is like if we expected primary school students to understand statements like the one above about adding odd and even numbers _without_ spending the time doing basic addition problems!  Furthermore, the solution cannot be asking you to perform the complicated process multiple times. Solving a matrix multiplication problem by hand takes ages.  Life is far too short to do 50 matrix multiplication problems by hand! 

This is why we think computer programming can be so useful to students learning mathematics. Computers can perform  complicated __processes__ such as matrix multiplication and drawing graphs quickly and accurately. You can use a computer to quickly apply the procedure many times and in doing so develop the __fluency__ that is required to understand general statements about the outcomes from these procedures.

# Calculating a matrix of angles between pairs of vectors (using NumPy)

Consider this equation that resembles equation 5 from the example report:

$$
\Theta = \arccos\left( \frac{\mathbf{X}^T\mathbf{X}}{\mathbf{m} \otimes \mathbf{m}^T} \right) \qquad \textrm{where} \qquad \mathbf{m} = \sqrt{ \left( \mathbf{X} \odot \mathbf{X} \right)^T \mathbf{1} }
$$

If $\mathbf{X}$ is the $2\times n$ dimensional array that was discussed in the last exercies then $\Theta$ is an $n\times n$, symmetric matrix in which all the diagonal elements are 0.  The off diagonal elements then contain the values of the $n(n-1)/2$ angles between the $n$ random vectors in $\mathbf{X}$.

We can understand why by starting with the numerator of the quotient, which is equal to the matrix of dot products that was discussed in the last exercise.  We then recall that the dot product of two vectors $\mathbf{a}$ and $\mathbf{b}$ is also given by:

$$
\mathbf{a} \cdot \mathbf{b} = |\mathbf{a}| |\mathbf{b} | \cos(\theta)
$$

Now notice that to compute the denominator in the quotient we first calculate the magnitudes of all the vectors in $X$ using:

$$
m = \sqrt{ \left(\mathbf{X}\odot \mathbf{X}\right)^T \mathbf{1} }
$$

$\mathbf{X}\odot \mathbf{X}$ indicates that we are taking the element-wise product of the matrix $\mathbf{X}$ with itself.  This is what is calculated if we do `X**2` on a NumPy array (If `X` is a 2D NumPy array then `X**n` __is not__ the $n$th power of the matrix!  To calculate the $n$th power of a symmetric matrix you use `np.linalg.matrix_power` which does all the matrix multiplication that is required to compute the matrix power.)  

$\mathbf{X}\odot \mathbf{X}$ is a $2 \times n$ dimensional matrix whose elements are equal to the squares of the elements of $\mathbf{X}$.  Consequently, if we transpose this matrix and multiply it by the vector $\mathbf{1} = (1,1)$ we get an $n$ dimensional vector that contains the square moduli of the vectors of the columns of $\mathbf{X}$.  The vector $\mathbf{m}$ that is defined above thus contains the moduli of all the vectors in the columns of $\mathbf{X}$.

If we now return to the denominator in the expression for $\Theta$ we see that we are computing the outer product of the vector $\mathbf{m}$ with itself.  This operation returns an $n\times n$ matrix in which element $i, j$ is the product of the moduli of the vectors in the $i$th and $j$th columns of $\mathbf{X}$.  Once we remember that the $i,j$th element of the matrix in the numerator of the quotient is the dot product of these two vectors we recognise that the quotient is the $n\times n$ matrix of arc cosines of the angles between the vectors in  $\mathbf{X}$ because $\mathbf{a} \cdot \mathbf{b} = |\mathbf{a}| |\mathbf{b} | \cos(\theta)$.

# Calculating the bounded power set of a set (by hand)

Suppose that you have a set, $S$, with a cardinality of 4, for example, $\{1,2,3,4\}$. The bounded power set $\mathcal{P}_2(S)$ is the set of all sets with cardinality 2 that can be constructed from $S$.  For our example this is:

$$
\mathcal{P}_2(S) = \{ \{2,1\}, \{3,1\}, \{4,1\}, \{3,2\}, \{4,2\}, \{4,3\} \}
$$

A useful way to think about how to construct this set is to construct the Cartesian product of the set with itself as I have done for our example set below:

$$
S \times S = \left[ 
\begin{matrix} 
(1,1) & (1,2) & (1,3) & (1,3) \\
(2,1) & (2,2) & (2,3) & (2,3) \\
(3,1) & (3,2) & (3,3) & (3,3) \\
(4,1) & (4,2) & (4,3) & (4,3) \\
\end{matrix}
\right]
$$

The elements of $S \times S$ are not the same as the elements of $\mathcal{P}_2(S)$. $S \times S$ has more elements than $\mathcal{P}_2(S)$ and the elements of $S \times S$ are ordered pairs and __not__ sets.  However, as you can see from the above example, if we convert the $\frac{n(n-1)}{2}$ ordered pairs that lie below the diagonal of the Cartesian product into sets we get the elements in $\mathcal{P}_2(S)$.

$$

# Conclusions

The exercises have shown that, if `A` and `B` are NumPy arrays with the same shape you can perform binary operations (addition, subtraction, multiplication, division, etc) element-wise using:

```python
sumAB = A + B
minusAB = A - B
prodAB = A*B
divAB = A / B
```

The `sumAB`, `minusAB`, `prodAB` and `divAB` variables that are created in the above code snippet are NumPy arrays with the same shape as `A` (if you use `np.divide( A, B )` instead of `A/B` you can even remove issues that arise when dividing by zero). 

It is important to realise that you are using `A` and `B` to store the vectors $\mathbf{a}$ and $\mathbf{b}$ `A*B` is __not__ the dot ($\mathbf{a}\cdot \mathbf{b}$ or $\mathbf{a}^T\mathbf{b}$) or cross product ($\mathbf{a} \times \mathbf{b}$) that appears in any mathematical derivation.  To compute the dot and cross products you need to use `np.dot` and `np.cross`. If by contrast `A` is storing a matrix and `B` is storing a vector then you will get an error if you write `A*B` in your code as `A` and `B` are not the same shape.  If you want to calculate the matrix vector product you need to use `np.matmul(A, B)`.  Similarly, if `A` and `B` are matrices `A*B` is __not__ the equal to the product of them that appears in most of your mathematical derivations.  To compute the matrix product using NumPy you again use `np.matmul`.

The most important thing to recognise, however, is that __there are NumPy implementations of every operation that you will learn to perform on vectors and matrices during your degree.__ Consequently, you should only ever be writing a function that, for instance, diagonalises a matrix as an exercise. In your production code you can use the implementation in NumPy. Trying to use these objects and functions that other people have written in the codes that you write is what (we believe) will provide you with the __fluency__ with the __objects__ that matheamtics concerns itself with that was discussed in the introduction. As a general rule, if you have written a complicated python program for doing a mathematical operation that doesn't use any matrices or NumPy arrays there is probably a better way of doing what you are doing.
